```python
!pip install pyECLAT

from pyECLAT import ECLAT

# Create ECLAT object
eclat = ECLAT(data=df)

# Get frequent itemsets
indices, support = eclat.fit(
    min_support=0.03,
    min_combination=2,
    max_combination=2
)

# Calculate Support, Confidence and Lift
rows = []

for itemset, sup_ab in support.items():

    items = itemset.split('&')
    A = items[0].strip()
    B = items[1].strip()

    sup_a = (df == A).any(axis=1).sum() / len(df)
    sup_b = (df == B).any(axis=1).sum() / len(df)

    confidence = sup_ab / sup_a
    lift = confidence / sup_b

    rows.append([
        itemset,
        round(sup_ab, 4),
        round(confidence, 4),
        round(lift, 4)
    ])

result_df = pd.DataFrame(
    rows,
    columns=['Itemset', 'Support', 'Confidence', 'Lift']
)

print(result_df)
```

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import fpgrowth, association_rules

# Load dataset
df = pd.read_csv('/content/Market_Basket_Optimisation.csv', header=None)

# Convert dataframe into transaction list
transactions = []

for i in range(len(df)):
    transactions.append(
        [str(item) for item in df.iloc[i].dropna().values]
    )

# Transaction Encoding
te = TransactionEncoder()
te_array = te.fit(transactions).transform(transactions)

df_encoded = pd.DataFrame(te_array, columns=te.columns_)

# Apply FP-Growth
frequent_itemsets = fpgrowth(
    df_encoded,
    min_support=0.03,
    use_colnames=True
)

print("Frequent Itemsets:")
print(frequent_itemsets)

# Generate Association Rules
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.2
)

# Display required columns
result = rules[['antecedents',
                'consequents',
                'support',
                'confidence',
                'lift']]

print("\nAssociation Rules:")
print(result)
```
